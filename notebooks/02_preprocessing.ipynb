```python
import pandas as pd
from feature_engineering import (
    encode_binary_columns, encode_label_columns, one_hot_encode_columns,
    scale_numeric_columns, create_tenure_group, add_total_services
)
df = pd.read_csv('../data/raw/telco-customer-churn-ibm.csv')
```

```python
# Encode binary Yes/No columns
df_encoded = encode_binary_columns(df, ['Partner', 'Dependents', 'PhoneService', 'PaperlessBilling', 'Churn'])
```

```python
# Label encode multi-category columns
df_encoded = encode_label_columns(df_encoded, ['gender', 'InternetService', 'Contract', 'PaymentMethod'])
```

```python
# Optionally: One-hot encode if you prefer
df_encoded = one_hot_encode_columns(df_encoded, ['InternetService', 'Contract', 'PaymentMethod'])
```

```python
# Create tenure group
df_encoded = create_tenure_group(df_encoded)
```

```python
# Add total services count
df_encoded = add_total_services(df_encoded)
```

```python
# Convert TotalCharges to numeric, coerce errors to NaN
df_encoded['TotalCharges'] = pd.to_numeric(df_encoded['TotalCharges'], errors='coerce')

# Fill NaNs with median (or mean, or zero, as appropriate)
median_total = df_encoded['TotalCharges'].median()
df_encoded['TotalCharges'].fillna(median_total, inplace=True)

# Double-check there are no NaNs left
print(df_encoded[['tenure', 'MonthlyCharges', 'TotalCharges']].isna().sum())

# Now safely scale
df_encoded = scale_numeric_columns(df_encoded, ['tenure', 'MonthlyCharges', 'TotalCharges'])
```
